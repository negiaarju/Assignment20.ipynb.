1.  What is unsupervised learning in the context of machine learning?

**Unsupervised Learning** is a type of **machine learning** where the model learns from **data that has no labeled output**. This means the algorithm tries to **find patterns, structures, or relationships in the data on its own**. 🤖📊

### Key Idea

* In unsupervised learning, the dataset **does not contain target labels**.
* The model **analyzes the data and groups or organizes it based on similarities**.

### Example

Suppose you have customer data with information like **age, income, and buying behavior**, but **no category labels**.
An unsupervised algorithm can **group similar customers together** into clusters based on their behavior.

### Common Unsupervised Learning Techniques

1. **Clustering** – Groups similar data points together

   * Example: Customer segmentation
   * Algorithms: K-Means, Hierarchical Clustering

2. **Association** – Finds relationships between variables

   * Example: Market basket analysis (items frequently bought together)

3. **Dimensionality Reduction** – Reduces the number of features while keeping important information

   * Example: PCA (Principal Component Analysis)


2. How does K-Means clustering algorithm work?

 **K-Means clustering** is an **unsupervised learning algorithm** used to group similar data points into **K clusters**.

### How it works (short):

1. **Choose K** – Decide the number of clusters.
2. **Initialize centroids** – Randomly place K cluster centers.
3. **Assign points** – Each data point is assigned to the nearest centroid.
4. **Update centroids** – Recalculate the centroid of each cluster based on assigned points.
5. **Repeat** steps 3–4 until the centroids no longer change.


3. Explain the concept of a dendrogram in hierarchical clustering.

A **dendrogram** is a **tree-like diagram** used in **hierarchical clustering** to show how data points are grouped together step by step.

* Each **branch** represents a cluster.
* The **height of the branches** shows the distance or similarity between clusters.
* Clusters that join **lower in the diagram are more similar** than those that join higher.


4.  What is the main difference between K-Means and Hierarchical Clustering?

The main difference between **K-Means** and **Hierarchical Clustering** is how they form clusters.

* **K-Means:** Requires the number of clusters (**K**) to be specified in advance and groups data by repeatedly updating cluster centroids.
* **Hierarchical Clustering:** Does **not require a predefined number of clusters** and builds a hierarchy of clusters shown using a **dendrogram**.


5.  What are the advantages of DBSCAN over K-Means?

**Advantages of DBSCAN over K-Means:**

1. **No need to specify number of clusters** in advance.
2. **Can detect clusters of arbitrary shapes**, not just circular ones.
3. **Handles noise and outliers well** by marking them as separate points.
4. **Works better for datasets with uneven cluster sizes.**


6. When would you use Silhouette Score in clustering?

The **Silhouette Score** is used in clustering to **evaluate how well the data points are grouped into clusters**.

* It measures **how similar a data point is to its own cluster compared to other clusters**.
* The score ranges from **-1 to 1**.

**Interpretation:**

* **1** → Data point fits well in its cluster
* **0** → Data point lies between two clusters
* **-1** → Data point may be in the wrong cluster


7.  What are the limitations of Hierarchical Clustering?

**Limitations of Hierarchical Clustering:**

1. **Computationally expensive** for large datasets.
2. **High memory usage** when the dataset is big.
3. **Sensitive to noise and outliers.**
4. **Once clusters are merged or split, the decision cannot be undone.**


8. Why is feature scaling important in clustering algorithms like K-Means?

**Feature scaling** is important in **K-Means clustering** because the algorithm uses **distance (usually Euclidean distance)** to group data points.

* If features have **different scales**, variables with **larger values dominate the distance calculation**.
* This can lead to **incorrect clustering results**.


9. How does DBSCAN identify noise points?

**DBSCAN** identifies **noise points** using two parameters: **epsilon (ε)** and **minimum points (MinPts)**.

* A point is a **core point** if it has at least **MinPts neighbors within distance ε**.
* A **border point** is close to a core point but has fewer neighbors.
* A **noise point** is a point that **does not have enough neighbors within ε and is not reachable from any core point**.


10.  Define inertia in the context of K-Means.

**Inertia** in **K-Means clustering** is the **sum of squared distances between each data point and its nearest cluster centroid**.

* It measures **how tightly the data points are grouped within clusters**.
* **Lower inertia** means **better and more compact clusters**.


11. What is the elbow method in K-Means clustering?

The **Elbow Method** in **K-Means clustering** is used to **find the optimal number of clusters (K)**.

* Run K-Means with different values of **K**.
* Calculate the **inertia (within-cluster sum of squares)** for each K.
* Plot **K vs inertia**.
* The point where the graph forms an **“elbow”** (sharp bend) indicates the best number of clusters.


12. Describe the concept of "density" in DBSCAN.

In **DBSCAN**, **density** refers to the **number of data points within a specified distance (ε)** from a point.

* If a point has **at least MinPts neighbors within distance ε**, it is called a **core point**.
* Points close to a core point are **border points**.
* Points in **low-density regions** with very few neighbors are treated as **noise (outliers)**.


13.  Can hierarchical clustering be used on categorical data?

Yes, **hierarchical clustering can be used on categorical data**, but it requires an appropriate **distance or similarity measure**.

* Instead of Euclidean distance, measures like **Hamming distance** or **Jaccard similarity** are used.
* These measures compare **similarities between categorical values**.


14. What does a negative Silhouette Score indicate?

A **negative Silhouette Score** indicates that a **data point is likely assigned to the wrong cluster**.

* It means the point is **closer to points in another cluster than to its own cluster**.


15.  Explain the term "linkage criteria" in hierarchical clustering.

**Linkage criteria** in **hierarchical clustering** refers to the **method used to measure the distance between clusters** when deciding which clusters to merge.

Common types include:

* **Single linkage** – distance between the closest points of two clusters.
* **Complete linkage** – distance between the farthest points.
* **Average linkage** – average distance between all points of the clusters.


16. Why might K-Means clustering perform poorly on data with varying cluster sizes or densities?

**K-Means clustering** may perform poorly on data with **varying cluster sizes or densities** because it assumes that clusters are **similar in size and roughly spherical**.

* It assigns points based on **distance to the nearest centroid**, which may cause **large clusters to absorb points from smaller clusters**.
* It also struggles when clusters have **different densities or irregular shapes**.



17. What are the core parameters in DBSCAN, and how do they influence clustering?

The **core parameters in DBSCAN** are:

1. **Epsilon (ε)** – The maximum distance between two points to be considered neighbors.
2. **MinPts (Minimum Points)** – The minimum number of points required in a neighborhood to form a dense region.

**Influence on clustering:**

* **Large ε** → More points join clusters; clusters may merge.
* **Small ε** → Fewer neighbors; more points may be labeled as noise.
* **High MinPts** → Requires denser regions to form clusters.
* **Low MinPts** → Easier to form clusters, but may include noise.


18.  How does K-Means++ improve upon standard K-Means initialization?

**K-Means++** improves standard **K-Means initialization** by choosing **initial centroids more carefully** instead of selecting them randomly.

* The first centroid is chosen **randomly**.
* The remaining centroids are chosen **farther away from existing centroids** based on probability.

**Benefits:**

* Reduces the chance of **poor clustering**.
* **Faster convergence**.
* Produces **more accurate clusters**.


19.  What is agglomerative clustering?

**Agglomerative clustering** is a type of **hierarchical clustering** that follows a **bottom-up approach**.

* Each data point starts as **its own cluster**.
* The **closest clusters are merged step by step**.
* This process continues until **all points form one cluster or the desired number of clusters is reached**.


20. What makes Silhouette Score a better metric than just inertia for model evaluation?

The **Silhouette Score** is better than **inertia** for model evaluation because it measures **both cluster cohesion and separation**.

* **Inertia** only measures how close data points are to their own cluster center.
* **Silhouette Score** checks **how similar a point is to its own cluster compared to other clusters**.

**Result:** It gives a **more complete evaluation of clustering quality**.


```python
#  Generate synthetic data with 4 centers using make_blobs and apply K-Means clustering. Visualize using ascatter plot.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# Generate synthetic data with 4 centers
X, y = make_blobs(n_samples=300, centers=4, random_state=42)

# Apply K-Means clustering
kmeans = KMeans(n_clusters=4, random_state=42)
labels = kmeans.fit_predict(X)

# Get cluster centers
centers = kmeans.cluster_centers_

# Scatter plot of clustered data
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.scatter(centers[:, 0], centers[:, 1], c='red', marker='X', s=200)
plt.title("K-Means Clustering with 4 Centers")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
#  Create synthetic data using make_blobs and evaluate silhouette scores for k = 2 to 5. Display as a bar chart.

import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# 1. Create data
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=1.0, random_state=42)

# 2. Evaluate silhouette scores for k=2..5
scores = []
k_range = range(2, 6)
for k in k_range:
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    labels = km.fit_predict(X)
    scores.append(silhouette_score(X, labels))

# 3. Bar chart
plt.bar(list(map(str, k_range)), scores, color='skyblue')
plt.xlabel("Clusters (k)")
plt.ylabel("Score")
plt.title("Silhouette Scores")
plt.show()
```

```python
#  Load the Iris dataset and use hierarchical clustering to group data. Plot a dendrogram with average linkage.

import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
import pandas as pd

# 1. Load the Iris dataset
iris = load_iris()
# The features are stored in iris.data
X = iris.data
# Optional: Get target names to use as labels
species_names = iris.target_names[iris.target]

# Optional: Standardize the data, as clustering is distance-based and sensitive to scale
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. Perform hierarchical clustering with 'average' linkage
# The linkage function computes the linkage matrix
# We use the scaled data for clustering
linked = linkage(X_scaled, method='average')

# 3. Plot the dendrogram
plt.figure(figsize=(15, 8))
plt.title("Hierarchical Clustering Dendrogram (Average Linkage)")
plt.xlabel("Sample Index or (Cluster Size)")
plt.ylabel("Distance")

# Plot the dendrogram
dendrogram(
    linked,
    orientation='top',
    labels=species_names, # Use actual species names as labels
    distance_sort='descending',
    show_leaf_counts=True, # Show the number of points in contracted nodes
    leaf_rotation=90., # Rotate the x-axis labels
    leaf_font_size=8., # Font size for the labels
)

plt.show()
```

```python
#  Generate synthetic data with overlapping clusters using make_blobs, then apply K-Means and visualize with decision boundarie.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from matplotlib.colors import ListedColormap

# 1. Generate synthetic data with overlapping clusters
# Increase 'cluster_std' (e.g., to 1.8) to create overlapping clusters
X, y_true = make_blobs(n_samples=400, centers=4, cluster_std=1.8, random_state=101)

# 2. Apply K-Means clustering
# We specify the number of clusters we expect (4)
kmeans = KMeans(n_clusters=4, init='k-means++', n_init=10, random_state=0)
kmeans.fit(X)
y_kmeans = kmeans.predict(X)
centroids = kmeans.cluster_centers_

# 3. Visualize the data and decision boundary

# Create a meshgrid to plot the decision boundary
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                     np.arange(y_min, y_max, 0.02))

# Predict the cluster for each point in the meshgrid
Z = kmeans.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Define color maps
cmap_light = ListedColormap(['#FFAAAA', '#AAFFAA', '#AAAAFF', '#FAEBD7'])
cmap_bold = ListedColormap(['#FF0000', '#00FF00', '#0000FF', '#8B4513'])

plt.figure(figsize=(10, 7))

# Plot the decision boundaries
plt.contourf(xx, yy, Z, cmap=cmap_light, alpha=0.6)

# Plot the actual data points (colored by K-Means labels)
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, cmap=cmap_bold, s=50, edgecolor='k')

# Plot the centroids
plt.scatter(centroids[:, 0], centroids[:, 1], marker='*', s=300, c='red', edgecolor='black', label='Centroids')
plt.title("K-Means Clustering with Overlapping Clusters and Decision Boundary")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#  Load the Digits dataset and apply DBSCAN after reducing dimensions with t-SNE. Visualize the results.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.manifold import TSNE
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler

# 1. Load the Digits dataset
digits = load_digits()
# The data is already flattened into 64 features per image (8x8 pixels)
data = digits.data
target = digits.target # The true labels for comparison

# 2. Standardize the data (recommended for distance-based algorithms like DBSCAN)
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data)

# 3. Reduce dimensions with t-SNE to 2 components for visualization
# It's often recommended to reduce dimensions with PCA first for very high-dim data
# But here we apply t-SNE directly for simplicity and to follow the request.
tsne = TSNE(n_components=2, perplexity=30, random_state=42, init='pca', learning_rate='auto')
# 'init="pca"' is often recommended to speed up computation and suppress noise
data_embedded = tsne.fit_transform(data_scaled)

# 4. Apply DBSCAN clustering
# Optimal parameters (eps, min_samples) often require tuning.
# For demonstration, we use sample values.
dbscan = DBSCAN(eps=3, min_samples=5)
clusters = dbscan.fit_predict(data_embedded)
# -1 in clusters means noise points

# 5. Visualize the results
plt.figure(figsize=(10, 7))

# Scatter plot colored by the DBSCAN assigned clusters
# Use the 'clusters' array for coloring
plt.scatter(data_embedded[:, 0], data_embedded[:, 1], c=clusters, cmap='Spectral', s=10)

# Optional: Add titles and axis labels
plt.title('DBSCAN Clustering on t-SNE Reduced Digits Dataset')
plt.xlabel('t-SNE component 1')
plt.ylabel('t-SNE component 2')
plt.colorbar(label='Cluster Label')

# Show the plot
plt.show()

# Print the number of clusters found (excluding noise, label -1)
n_clusters = len(set(clusters)) - (1 if -1 in clusters else 0)
n_noise = list(clusters).count(-1)
print(f"Estimated number of clusters: {n_clusters}")
print(f"Estimated number of noise points: {n_noise}")
```

```python
# Load the Breast Cancer dataset and compare inertia values for K = 2 to 6 using K-Means. Show results in a line plot.

import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Load the dataset
cancer = load_breast_cancer()
X = cancer.data

# 2. Standardize the data (important for K-Means)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Calculate inertia for K = 2 to 6
inertia_values = []
k_range = range(2, 7)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia_values.append(kmeans.inertia_)

# 4. Plot the results
plt.figure(figsize=(8, 5))
plt.plot(k_range, inertia_values, 'bo-')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method For Optimal K (Breast Cancer Dataset)')
plt.grid(True)
plt.show()
```

```python
#  Use the Wine dataset, apply DBSCAN after scaling the data, and count the number of clusters (excludingnoise).

import numpy as np
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
import pandas as pd

# 1. Load the Wine dataset
wine_data = load_wine()
X = wine_data.data

# 2. Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Apply DBSCAN
# Optimal parameters (eps and min_samples) for DBSCAN often require tuning
# using methods like a k-distance graph.
# For this example, we use common parameters; the result may vary with different values.
# Let's use eps=1.0 and min_samples=5 as an example.
dbscan = DBSCAN(eps=1.0, min_samples=5)
clusters = dbscan.fit_predict(X_scaled)
labels = dbscan.labels_ # Noise points are labeled as -1

# 4. Count the number of clusters (excluding noise)
# Convert labels to a set to find unique cluster IDs
unique_labels = set(labels)

# Count the number of clusters, excluding the noise label (-1) if present
n_clusters_ = len(unique_labels) - (1 if -1 in unique_labels else 0)

# 5. Print the result
print(f"Estimated number of clusters (excluding noise): {n_clusters_}")
print(f"Estimated number of noise points: {list(labels).count(-1)}")
```

```python
# Generate synthetic data with make_blobs and apply KMeans. Then plot the cluster centers on top of the data points.

import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import numpy as np

# 1. Generate synthetic data
# X contains the data points, y_true contains the actual (true) cluster labels
X, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

# 2. Apply KMeans clustering
# Initialize the KMeans model with the desired number of clusters (k=4)
# Suppress the warning for n_init with the specific parameter setting
kmeans = KMeans(n_clusters=4, init='k-means++', n_init=10, random_state=0)

# Fit the model to the data and predict the cluster labels for each point
y_kmeans = kmeans.fit_predict(X)

# Get the coordinates of the cluster centers
centers = kmeans.cluster_centers_

# 3. Plot the data points and cluster centers
plt.figure(figsize=(8, 6))

# Plot the data points, colored by their assigned cluster
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap='viridis', label='Data points')

# Plot the cluster centers (centroids)
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.7, marker='o', edgecolors='black', label='Cluster centers')

plt.title('K-Means Clustering with Cluster Centers')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Load the Iris dataset, cluster with DBSCAN, and print how many samples were identified as noise.

import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

# 1. Load the Iris dataset
# The dataset is a Bunch object, we access the feature data using the 'data' attribute
iris = load_iris()
X = iris.data
# It is often beneficial to scale the data when using distance-based algorithms
X = StandardScaler().fit_transform(X)

# 2. Cluster with DBSCAN
# Default parameters for DBSCAN are eps=0.5 and min_samples=5
# These parameters can be tuned for better results on a specific dataset
db = DBSCAN(eps=0.5, min_samples=5).fit(X)
labels = db.labels_

# 3. Count how many samples were identified as noise
# Noise points are assigned the label -1 in the labels_ attribute
n_noise_ = list(labels).count(-1)

print(f"Total samples in the dataset: {len(X)}")
print(f"Estimated number of noise points: {n_noise_}")
```

```python
# Generate synthetic non-linearly separable data using make_moons, apply K-Means, and visualize the clustering result.

import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans
import numpy as np

# 1. Generate Synthetic Data (non-linearly separable)
# make_moons creates two interleaving half-circles
X, y = make_moons(n_samples=300, noise=0.05, random_state=42)

# 2. Apply K-Means
# We set n_clusters=2 because we know there are two moons
kmeans = KMeans(n_clusters=2, random_state=42, n_init='auto')
labels = kmeans.fit_predict(X)

# 3. Visualize the Results
plt.figure(figsize=(10, 6))

# Plotting the data points colored by their K-Means cluster label
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=50, edgecolors='k', alpha=0.7)

# Plot the centroids discovered by K-Means
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, marker='X', label='Centroids')

plt.title("K-Means Clustering on Non-Linear 'Make Moons' Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#  Load the Digits dataset, apply PCA to reduce to 3 components, then use KMeans and visualize with a 3D scatter plot.

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn import datasets
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

# 1. Load the Digits dataset
digits = datasets.load_digits()
X = digits.data
y = digits.target # The actual target values can be used for comparison

# 2. Apply PCA to reduce to 3 components
# Standardizing data is generally recommended for PCA, but the digits data (0-16 pixel values) is already somewhat standardized.
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X)

# 3. Use K-Means
# The digits dataset has 10 actual classes (0-9), so we set n_clusters=10
kmeans = KMeans(n_clusters=10, random_state=42, n_init=10)
kmeans.fit(X_pca)
labels = kmeans.labels_

# 4. Visualize with a 3D Scatter plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Scatter plot, coloring points by their assigned K-Means cluster label
scatter = ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2], c=labels, cmap='viridis', s=50)

# Add labels and title
ax.set_xlabel('Principal Component 1')
ax.set_ylabel('Principal Component 2')
ax.set_zlabel('Principal Component 3')
ax.set_title('K-Means Clustering on Digits Dataset (PCA-reduced to 3D)')

# Add a color bar/legend
legend1 = ax.legend(*scatter.legend_elements(), title="Clusters")
ax.add_artist(legend1)

plt.show()
```

```python
# Generate synthetic blobs with 5 centers and apply KMeans. Then use silhouette_score to evaluate the clustering.

import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# 1. Generate synthetic blobs with 5 centers
# X contains the samples, y contains the true cluster labels (which are not used for clustering)
X, y = make_blobs(n_samples=500, centers=5, n_features=2, random_state=42) #

# Optional: Visualize the generated data
# plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', marker='o')
# plt.title("Synthetic Data with 5 Centers")
# plt.xlabel("Feature 1")
# plt.ylabel("Feature 2")
# plt.show()

# 2. Apply KMeans clustering
# Set n_clusters to 5, as the data was generated with 5 centers.
kmeans = KMeans(n_clusters=5, random_state=42, n_init="auto") #
kmeans.fit(X)
y_kmeans = kmeans.predict(X) # Predict the cluster labels for each data point

# 3. Use silhouette_score to evaluate the clustering
score = silhouette_score(X, y_kmeans) #

print(f"Silhouette Score: {score}")
```

```python
#  Load the Breast Cancer dataset, reduce dimensionality using PCA, and apply Agglomerative Clustering.Visualize in 2D

import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering

# 1. Load Dataset
data = load_breast_cancer()
X = data.data

# 2. Scale Data (Important for PCA)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Apply PCA (Reduce to 2 Dimensions)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# 4. Apply Agglomerative Clustering
cluster = AgglomerativeClustering(n_clusters=2, linkage='ward')
cluster_labels = cluster.fit_predict(X_pca)

# 5. Visualize in 2D
plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=cluster_labels, cmap='viridis', marker='o', edgecolor='k', s=50)
plt.title('Agglomerative Clustering of Breast Cancer Dataset (PCA Reduced)')
plt.xlabel('First Principal Component')
plt.ylabel('Second Principal Component')
plt.colorbar(label='Cluster Label')
plt.grid(True)
plt.show()
```

```python
#  Generate noisy circular data using make_circles and visualize clustering results from KMeans and DBSCAN side-by-side.

import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import DBSCAN, KMeans
from sklearn.datasets import make_circles

# 1. Generate noisy circular data
# X contains the features, y contains the true labels (0 for inner circle, 1 for outer)
X, y = make_circles(n_samples=500, shuffle=True, noise=0.1, random_state=42, factor=0.5)

# 2. Apply KMeans clustering
# K-Means often struggles with non-linearly separable data like concentric circles
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10) # Set n_init to avoid warning
kmeans_labels = kmeans.fit_predict(X)

# 3. Apply DBSCAN clustering
# DBSCAN is density-based and can find arbitrarily shaped clusters
# Optimal parameters (eps, min_samples) depend on the data density.
# For this data, eps=0.3 and min_samples=10 work well
dbscan = DBSCAN(eps=0.3, min_samples=10)
dbscan_labels = dbscan.fit_predict(X)

# 4. Visualize the results side-by-side
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Original data (optional, colored by true labels)
axes[0].scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', s=40)
axes[0].set_title("Original Data (True Labels)")

# KMeans results
# K-means assigns all points to a cluster, often misclassifying the circular data
axes[1].scatter(X[:, 0], X[:, 1], c=kmeans_labels, cmap='viridis', s=40)
axes[1].set_title(f"KMeans Clustering (n_clusters=2)")

# DBSCAN results
# DBSCAN correctly identifies the two circles and treats some sparse points as noise (label -1)
axes[2].scatter(X[:, 0], X[:, 1], c=dbscan_labels, cmap='viridis', s=40)
axes[2].set_title(f"DBSCAN Clustering (eps=0.3, min_samples=10)")

for ax in axes:
    ax.set_xlabel("X1")
    ax.set_ylabel("X2")

plt.tight_layout()
plt.show()
```

```python
# Generate synthetic data using make_blobs and apply Agglomerative Clustering with 'average' linkage.Visualize clusters

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering

# 1. Generate synthetic data using make_blobs
X, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)
# X contains the features, y_true contains the actual labels (optional, for comparison)

# Visualize the original generated data (optional)
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.title("Original Generated Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

# 2. Apply Agglomerative Clustering with 'average' linkage
# Specify the number of clusters (e.g., 4, matching the data generation) and the linkage method
agg_clustering = AgglomerativeClustering(n_clusters=4, linkage='average', metric='euclidean')

# Fit the model and predict the cluster labels
# The fit_predict method performs both steps
y_pred = agg_clustering.fit_predict(X)

# 3. Visualize the clustered data
plt.subplot(1, 2, 2)
# Use the predicted labels to color the points
plt.scatter(X[:, 0], X[:, 1], c=y_pred, s=50, cmap='viridis')
plt.title("Agglomerative Clustering with Average Linkage")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.tight_layout()
plt.show()
```

```python
#  Load the Wine dataset, apply KMeans, and visualize the cluster assignments in a seaborn pairplot (first 4features).

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Load the Wine dataset
wine_data = load_wine()
# Convert to a pandas DataFrame for easier manipulation and visualization
df = pd.DataFrame(data=wine_data.data, columns=wine_data.feature_names)

# 2. Apply KMeans clustering
# It's recommended to scale data for K-Means as it's distance-based
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# The wine dataset has 3 actual classes, so we set n_clusters=3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10) # Set n_init explicitly to suppress potential warnings
kmeans.fit(scaled_data)

# Add the cluster labels to the original DataFrame
df['cluster_assignment'] = kmeans.labels_

# Optional: Add the actual target labels for comparison in the plot (though KMeans is unsupervised)
df['actual_target'] = wine_data.target_names[wine_data.target]

# 3. Visualize the cluster assignments in a seaborn pairplot (first 4 features)
# Select the first four feature columns for the pairplot
selected_features = df.columns[:4]
# Include the 'cluster_assignment' column in the DataFrame for hue mapping
plot_df = df[selected_features.tolist() + ['cluster_assignment']]

# Create the pairplot, using the cluster assignments for color (hue)
# The 'hue' variable must be categorical for correct plotting in pairplot
sns.pairplot(plot_df, hue='cluster_assignment', palette='tab10', diag_kind='kde')
plt.suptitle('K-Means Cluster Assignments in Pairplot (First 4 Features)', y=1.02) # Add a title
plt.show()
```

```python
# Generate noisy blobs using make_blobs and use DBSCAN to identify both clusters and noise points. Print thecount.

import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# 1. Generate noisy blobs
# You can increase cluster_std to create more overlap and potential noise,
# or decrease it for denser, more defined clusters.
centers = [[1, 1], [-1, -1], [1, -1]]
X, labels_true = make_blobs(n_samples=750, centers=centers, cluster_std=0.4,
                            random_state=0)

X = StandardScaler().fit_transform(X)

# 2. Apply DBSCAN
# Tune eps and min_samples based on your data's density.
# eps: maximum distance between two samples to be considered in the same neighborhood.
# min_samples: minimum number of points in a neighborhood for a point to be considered a core point.
db = DBSCAN(eps=0.3, min_samples=10).fit(X)
labels = db.labels_

# 3. Count clusters and noise points
# Number of clusters in labels, ignoring noise if present
n_clusters_ = len(set(labels)) - (1 if -1 in labels else 0)
# Number of noise points (labeled as -1)
n_noise_ = list(labels).count(-1)

print(f"Estimated number of clusters: {n_clusters_}")
print(f"Estimated number of noise points: {n_noise_}")

# Optional: Visualization (requires matplotlib)
# Plot the results to visually confirm the counts.
unique_labels = set(labels)
colors = plt.cm.Spectral(np.linspace(0, 1, len(unique_labels)))

for k, col in zip(unique_labels, colors):
    if k == -1:
        # Black used for noise.
        col = 'k'

    class_member_mask = (labels == k)

    xy = X[class_member_mask]
    plt.plot(xy[:, 0], xy[:, 1], 'o', markerfacecolor=col,
             markeredgecolor='k', markersize=6)

plt.title(f'Estimated number of clusters: {n_clusters_}')
plt.show()
```

```python
# Load the Digits dataset, reduce dimensions using t-SNE, then apply Agglomerative Clustering and plot the clusters.

import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.manifold import TSNE
from sklearn.cluster import AgglomerativeClustering

# 1. Load the Digits dataset
digits = datasets.load_digits()
X = digits.data
y = digits.target # The true labels are useful for coloring the plot later

# 2. Reduce dimensions using t-SNE
# It is highly recommended to use PCA to reduce dimensions to a reasonable amount (e.g., 50) first for t-SNE
from sklearn.decomposition import PCA
pca = PCA(n_components=50)
X_pca = pca.fit_transform(X)

# Apply t-SNE on the PCA-reduced data
tsne = TSNE(n_components=2, init='pca', random_state=42, perplexity=30)
X_tsne = tsne.fit_transform(X_pca)

# 3. Apply Agglomerative Clustering
# We'll use 10 clusters, corresponding to the 10 digits (0-9)
agg_cluster = AgglomerativeClustering(n_clusters=10, linkage='ward')
# 'fit_predict' returns the cluster labels for each sample
cluster_labels = agg_cluster.fit_predict(X_tsne)

# 4. Plot the clusters
plt.figure(figsize=(10, 8))
# Scatter plot, colored by the assigned cluster labels
scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=cluster_labels, cmap='viridis', s=50, alpha=0.8)
plt.colorbar(scatter, ticks=range(10))
plt.title('Agglomerative Clustering on t-SNE reduced Digits Dataset')
plt.xlabel('t-SNE component 1')
plt.ylabel('t-SNE component 2')
plt.axis('off') # Turn off axis for a cleaner visualization
plt.show()

# Optional: Plot colored by true labels for comparison
plt.figure(figsize=(10, 8))
scatter_true = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap='viridis', s=50, alpha=0.8)
plt.colorbar(scatter_true, ticks=range(10))
plt.title('True Labels of Digits Dataset after t-SNE reduction')
plt.xlabel('t-SNE component 1')
plt.ylabel('t-SNE component 2')
plt.axis('off')
plt.show()
```
